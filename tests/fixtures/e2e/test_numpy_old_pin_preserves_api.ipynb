```python
# Pinned to the last numpy release before numpy.bool/numpy.int/numpy.float
# were hard-removed in 1.24. This is a real, documented API break, not a
# synthetic one - the point of this fixture is proving that correct pinning
# actually preserves old behavior, not just that some install happens.
#
# SABOTAGE TEST: to confirm this fixture can actually fail, drop the
# '==1.23.5' pin here AND from the matching pre-install line in
# Build-DockerCmd's python3.11 case in run_suite.ps1, then rerun. Cell 2's
# assertion should fail with AttributeError against latest numpy. Revert
# both afterward.
%pip install numpy==1.23.5
```

```python
import numpy
```

```python
# The only thing that gates pass/fail here is the actual API behavior -
# not a version-string check, which would fail for the wrong reason (a
# version mismatch) even if numpy.bool happened to still work, and would
# mask an AttributeError behind a less meaningful assertion during a
# sabotage test against an unpinned/latest numpy.
import warnings

with warnings.catch_warnings():
    warnings.simplefilter('ignore', DeprecationWarning)
    old_style_flag = numpy.bool(1)

assert old_style_flag is True, 'numpy.bool did not behave as the old bool alias'
print(f'numpy=={numpy.__version__}: numpy.bool alias still works as expected')
```
