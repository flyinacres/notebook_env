```python
# humanize is pre-installed matching this exact pin (already-satisfied path).
# tabulate is pre-installed too, but this line explicitly re-pins it to a
# version that does not exist on PyPI: the sequential engine attempts this
# pin, fails on it, and leaves the pre-existing tabulate install untouched -
# so the failure is silent unless downstream code actually checks for it.
%pip install humanize==4.16.0
%pip install tabulate==0.0.0.nonexistent
```

```python
# Both imports succeed: humanize installed cleanly, and tabulate's stale
# pre-existing install was never removed by the failed re-pin attempt.
import humanize
import tabulate
```

```python
# Downstream code that actually depends on the re-pin having succeeded
# must be able to detect that it silently didn't - this is the property
# under test, not the import itself.
import importlib.metadata

expected_pin = '0.0.0.nonexistent'
t_ver = importlib.metadata.version('tabulate')
assert t_ver == expected_pin, (
    f'tabulate re-pin to {expected_pin} failed silently: '
    f'still running pre-existing {t_ver}'
)
```
