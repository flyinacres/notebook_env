```python
# humanize is pre-installed matching this exact pin, so it takes the
# already-satisfied path. tabulate is pre-installed too, but this line
# explicitly overrides its pin to a version that does not exist on PyPI:
# the sequential engine must attempt this pin, fail on it specifically,
# and leave the pre-existing tabulate installation intact and importable.
%pip install humanize==4.16.0
%pip install tabulate==0.0.0.nonexistent
```

```python
# tabulate's re-pin attempt above failed, but its pre-existing install
# was never removed, so it remains importable here.
import humanize
import tabulate
```

```python
# Verify that the valid package installed cleanly and the package with
# a failed re-pin attempt is still present and functional
import importlib.metadata

h_ver = importlib.metadata.version('humanize')
t_ver = importlib.metadata.version('tabulate')

assert h_ver is not None and len(h_ver) > 0, 'humanize was not installed'
assert t_ver is not None and len(t_ver) > 0, 'tabulate was not installed'

table_data = [['Status', 'Package'], ['Installed', 'humanize'], ['Installed', 'tabulate']]
rendered = tabulate.tabulate(table_data, headers='firstrow')
print(rendered)
print(f'Partial install succeeded for valid packages: humanize=={h_ver}, tabulate=={t_ver}')
```
