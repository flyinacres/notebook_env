```python
# Import and functional smoke test
import local_test_pkg
import importlib.metadata

# 1. Verify metadata exists and is loaded
dist_ver = importlib.metadata.version('local_test_pkg')
assert dist_ver is not None and len(dist_ver) > 0, 'local_test_pkg version metadata not found'

# 2. Functional check: the installed code's own reported version must match
# the installed metadata's version. No hardcoded expected literal here -- this
# same notebook is regenerated and re-executed against whichever version was
# pre-installed before generation, so it proves the pin-and-install mechanism
# reproduces *whatever* version it claims to, not one specific number.
func_ver = local_test_pkg.which_version()
assert func_ver == dist_ver, f'functional version {func_ver} does not match installed metadata version {dist_ver}'
print(f'Pinned install verification passed: local_test_pkg {dist_ver} active')
```
