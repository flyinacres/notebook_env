```python
import sys, subprocess, tempfile
cmd = [sys.executable, '-m', 'pip', 'install', '--no-input', '--disable-pip-version-check', 'boltons==24.0.0']
with tempfile.TemporaryFile() as out_f:
    res = subprocess.run(cmd, stdin=subprocess.DEVNULL, stdout=out_f, stderr=out_f, timeout=120)
assert res.returncode == 0, f'Install failed with rc {res.returncode}'
import boltons
print('Real install succeeded! boltons imported cleanly.')
```
